O programa solicita um número e cria a variável divisores para contar quantas divisões exatas ele possui. O for percorre todos os números de 1 até o valor informado e utiliza o operador % para verificar se existe resto na divisão. Quando o resto é zero, significa que aquele valor é um divisor e a variável divisores aumenta em 1.

No final, o programa verifica se foram encontrados exatamente dois divisores. Esse método foi escolhido porque um número primo só pode ser dividido exatamente por 1 e por ele mesmo. Assim, em vez de tentar identificar diretamente se o número é primo, o programa analisa uma característica que todo número primo obrigatoriamente possui.

20. Dada uma matriz (lista de listas), crie sua matriz transposta.

```python
matriz = [  [1, 2, 3], [4, 5, 6]]  # cria a matriz original

transposta = []  # cria uma lista vazia para armazenar a matriz transposta

for coluna in range(len(matriz[0])):  # percorre cada coluna da matriz
    nova_linha = []  # cria uma nova linha vazia

    for linha in range(len(matriz)):  # percorre cada linha da matriz
        nova_linha.append(matriz[linha][coluna])  # adiciona os elementos da coluna na nova linha

    transposta.append(nova_linha)  # adiciona a nova linha na matriz transposta

print(transposta)  # mostra a matriz transposta
```

O programa começa com uma matriz já pronta e cria uma lista vazia para guardar a matriz transposta. Depois, são usados dois laços for: um para percorrer as colunas e outro para percorrer as linhas.

Esse método foi escolhido porque, para formar a matriz transposta, é necessário trocar as linhas pelas colunas. Por isso, cada coluna da matriz original é percorrida e transformada em uma nova linha. Dessa forma, o programa consegue reorganizar todos os elementos automaticamente, sem precisar montar a nova matriz manualmente.

21. Implemente a busca binária para encontrar um elemento em uma lista ordenada.

```python
def busca_binaria(lista, numero):  # cria uma função para procurar um número na lista
    inicio = 0  # define o início da busca na primeira posição da lista
    fim = len(lista) - 1  # define o fim da busca na última posição da lista

    while inicio <= fim:  # repete enquanto ainda houver posições para verificar
        meio = (inicio + fim) // 2  # calcula a posição do meio da lista

        if lista[meio] == numero:  # verifica se o número está na posição do meio
            return meio  # retorna a posição onde o número foi encontrado

        elif numero > lista[meio]:  # verifica se o número procurado é maior que o valor do meio
            inicio = meio + 1  # passa a procurar somente na metade direita da lista

        else:  # caso o número seja menor que o valor do meio
            fim = meio - 1  # passa a procurar somente na metade esquerda da lista

    return -1  # retorna -1 caso o número não seja encontrado


lista = [2, 5, 8, 10, 14, 20, 25]  # cria uma lista ordenada de números

numero = int(input("Digite o número que deseja procurar: "))  # solicita e armazena o número fornecido pelo usuário

resultado = busca_binaria(lista, numero)  # chama a função e armazena o resultado da busca

if resultado != -1:  # verifica se o número foi encontrado
    print("Número encontrado na posição:", resultado)  # mostra a posição do número
else:
    print("Número não encontrado.")  # informa que o número não está na lista
```

O programa cria uma função chamada busca_binaria para procurar um número dentro de uma lista ordenada. Primeiro, são definidas as variáveis inicio e fim, que representam os limites da parte da lista que ainda precisa ser analisada.

Dentro do while, o programa calcula a posição do meio da lista. Se o número procurado estiver nessa posição, a função retorna sua localização. Caso o número seja maior que o valor do meio, a busca continua apenas na metade direita. Se for menor, continua somente na metade esquerda. O processo se repete até o número ser encontrado ou não existirem mais posições para verificar.

Esse método foi escolhido porque a lista já está ordenada. Isso permite eliminar metade dos valores a cada comparação, em vez de verificar todos os elementos um por um. Por isso, a busca binária é mais eficiente principalmente quando a lista possui muitos elementos. O retorno -1 foi utilizado para indicar que o número não foi encontrado, facilitando a verificação feita no final do programa.


22. Crie uma função para criptografar uma mensagem usando a cifra de César.

```python
def criptografar(mensagem, deslocamento):  # cria uma função para criptografar a mensagem
    alfabeto = "abcdefghijklmnopqrstuvwxyz"  # armazena todas as letras do alfabeto
    resultado = ""  # cria uma variável vazia para armazenar a mensagem criptografada

    for letra in mensagem:  # percorre cada letra da mensagem
        if letra in alfabeto:  # verifica se a letra está no alfabeto
            posicao = alfabeto.index(letra)  # identifica a posição da letra no alfabeto
            nova_posicao = (posicao + deslocamento) % 26  # calcula a nova posição da letra
            resultado += alfabeto[nova_posicao]  # adiciona a nova letra ao resultado

        else:  # caso o caractere não esteja no alfabeto
            resultado += letra  # mantém o caractere original

    return resultado  # retorna a mensagem criptografada


mensagem = input("Digite uma mensagem: ")  # solicita e armazena a mensagem fornecida pelo usuário
deslocamento = int(input("Digite o deslocamento: "))  # solicita e armazena o valor do deslocamento

mensagem_criptografada = criptografar(mensagem, deslocamento)  # chama a função e armazena a mensagem criptografada

print("Mensagem criptografada:", mensagem_criptografada)  # mostra a mensagem criptografada
```

O programa cria uma função para criptografar uma mensagem usando a Cifra de César. Primeiro, é definido o alfabeto e uma variável vazia chamada resultado, que será usada para montar a nova mensagem.

O for percorre cada letra da mensagem. Quando a letra está no alfabeto, o programa identifica sua posição e soma o valor do deslocamento escolhido pelo usuário. O uso de % 26 foi feito para garantir que, ao passar da letra z, a contagem volte para o início do alfabeto. Por exemplo, com deslocamento 3, a letra z continua a contagem a partir do começo.

Esse método foi escolhido porque a Cifra de César funciona justamente deslocando cada letra uma quantidade fixa de posições no alfabeto. O programa também mantém espaços, números e outros caracteres sem alteração, já que eles não fazem parte do alfabeto definido. Isso permite criptografar frases completas sem perder a estrutura original da mensagem.

23. Crie uma função para descriptografar uma mensagem cifrada com a cifra de César

```python
def descriptografar(mensagem, deslocamento):  # cria uma função para descriptografar a mensagem
    alfabeto = "abcdefghijklmnopqrstuvwxyz"  # armazena todas as letras do alfabeto
    resultado = ""  # cria uma variável vazia para armazenar a mensagem descriptografada

    for letra in mensagem:  # percorre cada letra da mensagem
        if letra in alfabeto:  # verifica se a letra está no alfabeto
            posicao = alfabeto.index(letra)  # identifica a posição da letra no alfabeto
            nova_posicao = (posicao - deslocamento) % 26  # calcula a posição original da letra
            resultado += alfabeto[nova_posicao]  # adiciona a letra descriptografada ao resultado

        else:  # caso o caractere não esteja no alfabeto
            resultado += letra  # mantém o caractere original

    return resultado  # retorna a mensagem descriptografada


mensagem = input("Digite a mensagem criptografada: ")  # solicita e armazena a mensagem fornecida pelo usuário
deslocamento = int(input("Digite o deslocamento: "))  # solicita e armazena o valor do deslocamento

mensagem_original = descriptografar(mensagem, deslocamento)  # chama a função e armazena a mensagem original

print("Mensagem descriptografada:", mensagem_original)  # mostra a mensagem descriptografada
```

```python
def validar_cpf(cpf):  # cria uma função para validar o CPF

    cpf_limpo = cpf.replace(".", "").replace("-", "")  # remove pontos e traço do CPF

    if len(cpf_limpo) != 11:  # verifica se o CPF possui 11 números
        return False

    elif cpf_limpo == cpf_limpo[0] * 11:  # verifica se todos os números são iguais
        return False

    soma_primeiro = 0  # armazena a soma do primeiro dígito

    for i in range(9):  # percorre os 9 primeiros números
        soma_primeiro += int(cpf_limpo[i]) * (10 - i)

    primeiro_digito = (soma_primeiro * 10) % 11  # calcula o primeiro dígito

    if primeiro_digito == 10:  # verifica se o resultado foi 10
        primeiro_digito = 0

    if primeiro_digito != int(cpf_limpo[9]):  # compara com o dígito do CPF
        return False

    soma_segundo = 0  # armazena a soma do segundo dígito

    for i in range(10):  # percorre os 10 primeiros números
        soma_segundo += int(cpf_limpo[i]) * (11 - i)

    segundo_digito = (soma_segundo * 10) % 11  # calcula o segundo dígito

    if segundo_digito == 10:  # verifica se o resultado foi 10
        segundo_digito = 0

    if segundo_digito != int(cpf_limpo[10]):  # compara com o dígito do CPF
        return False

    return True


cpf = input("Digite um CPF: ")

if validar_cpf(cpf):
    print("CPF válido")
else:
    print("CPF inválido")
```

O programa cria uma função para validar um CPF e começa retirando os pontos e o traço para trabalhar apenas com os números. Em seguida, verifica se o CPF possui 11 dígitos e se todos eles são iguais, pois esses casos não são considerados válidos.

Depois, o programa calcula separadamente os dois dígitos verificadores. Para o primeiro, ele usa os nove primeiros números do CPF e multiplica cada um por pesos que vão de 10 até 2. Após isso, calcula o resultado e compara com o primeiro dígito verificador informado no CPF. O mesmo processo é feito para o segundo dígito, agora usando os dez primeiros números e pesos de 11 até 2.

Esse método foi escolhido porque um CPF não pode ser validado apenas pelo tamanho ou pelo formato. É necessário refazer o cálculo dos dígitos verificadores e comparar com os números presentes no CPF. As variáveis foram separadas entre primeiro e segundo cálculo para deixar cada etapa mais organizada e evitar confusão entre os valores. O elif foi usado nas verificações iniciais porque, se a primeira condição já for verdadeira, não é necessário testar a seguinte.

```python
import random  # importa a biblioteca random para sortear uma palavra

palavras = ["carro", "lapis", "vitor", "python", "seiva, alberto, albeto"]  # cria uma lista com as palavras do jogo

palavra = random.choice(palavras)  # escolhe uma palavra aleatória da lista

letras_acertadas = []  # cria uma lista vazia para armazenar as letras acertadas
tentativas = 6  # define a quantidade de tentativas do jogador

while tentativas > 0:  # repete o jogo enquanto ainda houver tentativas

    palavra_mostrada = ""  # cria uma variável para mostrar a palavra com letras e traços

    for letra in palavra:  # percorre cada letra da palavra sorteada
        if letra in letras_acertadas:  # verifica se a letra já foi acertada
            palavra_mostrada += letra  # mostra a letra acertada
        else:
            palavra_mostrada += "_"  # mostra um traço no lugar da letra ainda não acertada

    print("\nPalavra:", palavra_mostrada)  # mostra como está a palavra
    print("Tentativas restantes:", tentativas)  # mostra a quantidade de tentativas restantes

    if palavra_mostrada == palavra:  # verifica se todas as letras foram descobertas
        print("Você ganhou!")  # informa que o jogador venceu
        break  # encerra o jogo

    palpite = input("Digite uma letra: ").lower()  # solicita e armazena a letra fornecida pelo usuário

    if palpite in palavra:  # verifica se a letra está na palavra
        letras_acertadas.append(palpite)  # adiciona a letra na lista de letras acertadas
        print("Você acertou!")  # informa que o jogador acertou

    else:
        tentativas -= 1  # diminui uma tentativa
        print("Você errou!")  # informa que o jogador errou

if tentativas == 0:  # verifica se todas as tentativas acabaram
    print("Você perdeu!")  # informa que o jogador perdeu
    print("A palavra era:", palavra)  # mostra qual era a palavra correta
```

O programa cria um jogo da forca escolhendo uma palavra aleatória de uma lista com random.choice(). Depois, utiliza uma lista chamada letras_acertadas para guardar os palpites corretos e define seis tentativas para o jogador.

O while foi escolhido porque o jogo precisa continuar até acontecer uma das duas situações: o jogador descobrir a palavra ou acabar as tentativas. Dentro dele, o for percorre cada letra da palavra para montar palavra_mostrada, exibindo as letras já acertadas e colocando _ nas que ainda não foram descobertas.

Quando o jogador informa uma letra, o programa verifica se ela está na palavra. Se estiver, ela é adicionada à lista de acertos. Caso contrário, uma tentativa é retirada. O break é usado para encerrar o while assim que a palavra inteira for descoberta.

Esse método foi escolhido porque o jogo depende de várias rodadas e não é possível saber antecipadamente quantas serão necessárias. A combinação de while, for, if e listas permite controlar as tentativas, os acertos e a formação da palavra durante toda a partida.

31. Implemente o algoritmo de ordenação por bolha (Bubble Sort).

```python
numeros = list(map(int, input("Digite os números separados por espaço: ").split()))  # recebe os números digitados pelo usuário e os coloca em uma lista

for i in range(len(numeros)):  # controla a quantidade de vezes que a lista será percorrida
    for j in range(len(numeros) - 1 - i):  # percorre os números que ainda precisam ser ordenados

        if numeros[j] > numeros[j + 1]:  # verifica se o número atual é maior que o próximo
            numeros[j], numeros[j + 1] = numeros[j + 1], numeros[j]  # troca os números de posição

print("Lista ordenada:", numeros)  # mostra a lista em ordem crescente
```

O programa recebe vários números do usuário e os armazena em uma lista. Depois, utiliza dois laços for para comparar os elementos da lista várias vezes e organizar os valores em ordem crescente.

O segundo for compara cada número com o próximo. Quando o número atual é maior, os dois trocam de posição. A cada passagem, os maiores valores vão sendo levados para o final da lista.

Esse método foi escolhido porque representa o funcionamento do Bubble Sort, que ordena a lista por meio de comparações entre elementos vizinhos. O uso de dois for é necessário porque uma única passagem não garante que todos os números fiquem na posição correta. A parte len(numeros) - 1 - i também evita comparar novamente os valores que já foram organizados no final da lista.

```python
tabuleiro = [" ", " ", " ", " ", " ", " ", " ", " ", " "]  # cria o tabuleiro com 9 espaços vazios


def mostrar_tabuleiro():  # cria uma função para mostrar o tabuleiro
    print(tabuleiro[0], "|", tabuleiro[1], "|", tabuleiro[2])
    print("--+---+--")
    print(tabuleiro[3], "|", tabuleiro[4], "|", tabuleiro[5])
    print("--+---+--")
    print(tabuleiro[6], "|", tabuleiro[7], "|", tabuleiro[8])


def verificar_vencedor():  # cria uma função para verificar se algum jogador venceu

    combinacoes = [
        [0, 1, 2],
        [3, 4, 5],
        [6, 7, 8],
        [0, 3, 6],
        [1, 4, 7],
        [2, 5, 8],
        [0, 4, 8],
        [2, 4, 6]
    ]  # armazena todas as combinações possíveis para vencer

    for combinacao in combinacoes:  # percorre cada combinação
        a, b, c = combinacao

        if tabuleiro[a] == tabuleiro[b] == tabuleiro[c] != " ":  # verifica se existem 3 símbolos iguais
            return True

    return False


jogador = "X"  # define que o jogador X começa

for rodada in range(9):  # permite no máximo 9 jogadas

    mostrar_tabuleiro()  # mostra o tabuleiro

    posicao = int(input(f"Jogador {jogador}, escolha uma posição de 1 a 9: ")) - 1  # solicita a posição escolhida pelo jogador

    while tabuleiro[posicao] != " ":  # verifica se a posição já está ocupada
        print("Essa posição já está ocupada!")
        posicao = int(input("Escolha outra posição: ")) - 1

    tabuleiro[posicao] = jogador  # coloca o símbolo do jogador na posição escolhida

    if verificar_vencedor():  # verifica se o jogador venceu
        mostrar_tabuleiro()
        print(f"Jogador {jogador} venceu!")
        break

    if jogador == "X":  # troca o jogador
        jogador = "O"
    else:
        jogador = "X"

else:
    mostrar_tabuleiro()
    print("Empate!")
```

Depois de cada jogada, a função verificar_vencedor percorre as combinações e compara se existem três símbolos iguais em uma mesma linha, coluna ou diagonal.

Esse método foi escolhido porque separar a exibição do tabuleiro e a verificação da vitória em funções deixa o código mais organizado e evita repetir as mesmas instruções várias vezes. 37. Implemente uma função recursiva para calcular o fatorial de um número.

```python
def fatorial(numero):  # cria uma função para calcular o fatorial
    if numero == 0 or numero == 1:  # verifica se chegou ao caso básico
        return 1  # retorna 1 e encerra as chamadas

    else:
        return numero * fatorial(numero - 1)  # chama a própria função diminuindo 1 do número


numero = int(input("Digite um número: "))  # solicita e armazena o número fornecido pelo usuário

resultado = fatorial(numero)  # chama a função e armazena o resultado

print("Fatorial:", resultado)  # mostra o resultado do fatorial
```

O programa cria uma função chamada fatorial para calcular o fatorial de um número usando recursividade. Primeiro, ele verifica se o número chegou a 0 ou 1, pois nesses casos o resultado do fatorial é 1.

Esse método foi escolhido para mostrar o uso de recursividade, já que o cálculo do fatorial pode ser definido a partir do próprio fatorial do número anterior. O caso básico é importante para impedir que a função continue se chamando infinitamente.

46. Calcule a média e o desvio padrão de uma lista de números.

```python
import math  # importa a biblioteca math para usar a raiz quadrada

numeros = list(map(float, input("Digite os números separados por espaço: ").split()))  # recebe os números digitados pelo usuário e os coloca em uma lista

media = sum(numeros) / len(numeros)  # calcula a média dos números

soma = 0  # cria uma variável para armazenar a soma das diferenças

for numero in numeros:  # percorre cada número da lista
    soma += (numero - media) ** 2  # calcula a diferença para a média e eleva ao quadrado

variancia = soma / len(numeros)  # calcula a variância

desvio_padrao = math.sqrt(variancia)  # calcula a raiz quadrada da variância

print(f"Média: {media:.2f}")  # mostra a média com 2 casas decimais
print(f"Desvio padrão: {desvio_padrao:.2f}")  # mostra o desvio padrão com 2 casas decimais
```

Esses valores são somados para encontrar a variância e, em seguida, é calculado o desvio padrão. Para essa etapa, foi importada a biblioteca math, pois ela permite usar a função math.sqrt(), responsável por calcular a raiz quadrada da variância de forma simples.

48. Implemente o algoritmo de ordenação por inserção (Insertion Sort).

```python
numeros = list(map(int, input("Digite os números separados por espaço: ").split()))  # recebe os números digitados pelo usuário e os coloca em uma lista

for i in range(1, len(numeros)):  # percorre a lista a partir do segundo elemento
    atual = numeros[i]  # guarda o número que será inserido na posição correta
    j = i - 1  # começa a comparação com o número anterior

    while j >= 0 and numeros[j] > atual:  # verifica se os números anteriores são maiores
        numeros[j + 1] = numeros[j]  # desloca o número maior uma posição para a direita
        j -= 1  # volta uma posição na lista

    numeros[j + 1] = atual  # coloca o número atual na posição correta

print("Lista ordenada:", numeros)  # mostra a lista em ordem crescente
```

O programa recebe vários números digitados pelo usuário e os armazena em uma lista. Depois, cria um dicionário vazio chamado contagem, que será usado para guardar cada número junto com a quantidade de vezes que ele aparece.

O for percorre todos os números da lista. Se o número já estiver no dicionário, sua contagem aumenta em 1. Caso ainda não esteja, ele é adicionado com o valor inicial 1.

Esse método foi escolhido porque o dicionário permite relacionar diretamente cada número com sua quantidade de repetições. Assim, o programa consegue contar todos os valores sem precisar criar uma variável separada para cada número.
