2. Gere uma matriz 4x4 com numeros inteiros de 1 a 16 e extraia a ultima
linha.

```python
import numpy as np  # Importa a biblioteca NumPy e define o apelido np

matriz = np.arange(1, 17).reshape(4, 4)  # Cria números de 1 a 16 e organiza em uma matriz 4x4

ul = matriz[-1, :]  # Seleciona a última linha da matriz

print("Matriz 4x4:")  # Exibe uma mensagem
print(matriz)  # Exibe a matriz completa

print("Última linha:")  # Exibe uma mensagem
print(ul)  # Exibe somente a última linha
```

Primeiro, utilizei np.arange(1, 17) para gerar os números inteiros de 1 até 16. O valor 17 é usado como limite porque ele não é incluído na sequência.

Em seguida, utilizei .reshape(4, 4) para reorganizar esses 16 números em uma matriz com 4 linhas e 4 colunas. Para obter somente a última linha, foi utilizado matriz[-1, :]. O índice -1 representa a última linha da matriz e o : indica que devem ser selecionadas todas as colunas dessa linha.

Escolhi essa forma porque permite primeiro criar todos os valores em sequência e depois organizá-los no formato solicitado, além de utilizar a própria indexação do NumPy para acessar diretamente a última linha.

Para encontrar a moda, primeiro utilizei np.unique(array, return_counts=True). Essa função retorna os diferentes valores existentes no array e também informa quantas vezes cada um aparece. Depois, np.argmax(contagens) identifica a posição que possui a maior quantidade de repetições, permitindo descobrir qual é o valor que aparece mais vezes.

Depois, utilizei np.corrcoef(array1, array2) para calcular a correlação entre os dois conjuntos. A função retorna uma matriz de correlação, então o trecho [0, 1] seleciona o valor que representa a relação entre o primeiro e o segundo array.

O resultado da correlação varia de -1 a 1. Valores próximos de 1 indicam uma relação positiva, valores próximos de -1 indicam uma relação negativa e valores próximos de 0 indicam pouca relação linear entre os dois arrays.

Essa forma foi escolhida porque permite gerar os valores e calcular a correlação diretamente com funções do NumPy, mantendo o código simples e fácil de entender.

10. Crie uma matriz identidade 6x6 e substitua a diagonal principal por valores crescentes de 1 a 6.

```python
import numpy as np  # Importa a biblioteca NumPy e utiliza o apelido np

matriz = np.eye(6)  # Cria uma matriz identidade 6x6

matriz[0, 0] = 1  # Substitui o primeiro valor da diagonal por 1
matriz[1, 1] = 2  # Substitui o segundo valor da diagonal por 2
matriz[2, 2] = 3  # Substitui o terceiro valor da diagonal por 3
matriz[3, 3] = 4  # Substitui o quarto valor da diagonal por 4
matriz[4, 4] = 5  # Substitui o quinto valor da diagonal por 5
matriz[5, 5] = 6  # Substitui o sexto valor da diagonal por 6

print("Matriz final:")  # Exibe uma mensagem
print(matriz)  # Exibe a matriz com a diagonal modificada
```

Para criar a matriz, utilizei np.eye(6), porque essa função gera diretamente uma matriz identidade 6x6. Nesse tipo de matriz, os valores da diagonal principal são iguais a 1 e todos os outros elementos são iguais a 0.

Depois, alterei individualmente os elementos da diagonal principal. Esses elementos possuem o mesmo índice de linha e coluna, como [0, 0], [1, 1], [2, 2] e assim por diante. Por isso, foi possível substituir os valores originais pelos números crescentes de 1 até 6.

Escolhi fazer dessa forma porque deixa bem visível quais posições pertencem à diagonal principal e como cada uma delas está sendo modificada. Como a matriz possui apenas seis posições na diagonal, a alteração individual também mantém o código simples e fácil de entender.

19. Crie dois arrays de numeros inteiros aleatorios de tamanho 20 e calcule o
angulo entre eles (produto escalar normalizado).

```python
import numpy as np  # Importa a biblioteca NumPy e utiliza o apelido np

array1 = np.random.randint(1, 101, 20)  # Cria o primeiro array com 20 números inteiros aleatórios

array2 = np.random.randint(1, 201, 20)  # Cria o segundo array com 20 números inteiros aleatórios

escalar = np.dot(array1, array2)  # Calcula o produto escalar entre os dois arrays

norma1 = np.sqrt(np.sum(array1 ** 2))  # Calcula a norma do primeiro array

norma2 = np.sqrt(np.sum(array2 ** 2))  # Calcula a norma do segundo array

normalizado = escalar / (norma1 * norma2)  # Calcula o produto escalar normalizado

print("Array 1:", array1)  # Exibe o primeiro array
print("Array 2:", array2)  # Exibe o segundo array
print("Produto escalar normalizado:", normalizado)  # Exibe o resultado
```

Primeiro criei dois arrays com 20 números inteiros e calculei o produto escalar usando np.dot(). Depois calculei o tamanho de cada vetor usando a soma dos valores ao quadrado e a raiz quadrada. Por fim, dividi o produto escalar pelo produto dessas duas normas, obtendo o produto escalar normalizado, que representa a relação angular entre os vetores

1. Calcule a integral da funcao f(x) = x2 sin(x) no intervalo de 0 a π.

```python
from scipy import integrate  # Importa o módulo de integração do SciPy
import numpy as np  # Importa o NumPy para utilizar seno e pi

def f(x):  # Define a função matemática
    return x ** 2 * np.sin(x)  # Representa f(x) = x² * sen(x)

resultado, _ = integrate.quad(f, 0, np.pi)  # Calcula a integral de 0 até pi e ignora o erro estimado

print("Resultado da integral:", resultado)  # Exibe somente o resultado
```

10. Calcule a integral dupla da função f(x,y)=xy+x 2, no intervalo [0,2] para x e [0,1] para y.

```python
from scipy import integrate  # Importa o módulo de integração do SciPy

def f(x, y):  # Define a função de duas variáveis
    return x * y + x ** 2  # Representa f(x,y) = xy + x²

def integral_y(x):  # Define a primeira integração, mantendo x fixo
    resultado, _ = integrate.quad(lambda y: f(x, y), 0, 1)  # Integra em relação a y de 0 até 1
    return resultado  # Retorna o resultado dessa primeira integral

resultado, _ = integrate.quad(integral_y, 0, 2)  # Integra o resultado anterior em relação a x de 0 até 2

print("Resultado da integral dupla:", resultado)  # Exibe o resultado final
```

Como a função possui duas variáveis, a integração foi feita em duas etapas. Primeiro integrei f(x,y) em relação a y, no intervalo de 0 a 1, mantendo x fixo. Depois, o resultado dessa primeira integração foi integrado em relação a x, de 0 até 2.

Utilizei integrate.quad() nas duas etapas porque essa é a função de integração apresentada para o SciPy. O _ foi usado para ignorar o erro estimado retornado pelo quad(), já que o exercício pede apenas o resultado da integral. O valor final será aproximadamente 3.6667, equivalente a 11/3

15. Crie um DataFrame com dados faltantes e use a interpolacao linear para
preenche-los.

```python
import pandas as pd  # Importa a biblioteca Pandas
import numpy as np  # Importa a biblioteca NumPy
from scipy.interpolate import interp1d  # Importa a função de interpolação linear

dados = {  # Cria os dados que serão utilizados no DataFrame
    'X': [1, 2, 3, 4, 5],
    'Valor': [10, np.nan, 30, np.nan, 50]
}

df = pd.DataFrame(dados)  # Cria o DataFrame contendo valores faltantes

x = np.array([1, 3, 5])  # Define as posições que possuem valores conhecidos
y = np.array([10, 30, 50])  # Define os valores conhecidos

interpolacao = interp1d(x, y, kind='linear')  # Cria a função de interpolação linear

df.loc[1, 'Valor'] = interpolacao(2)  # Preenche o valor faltante correspondente a X = 2
df.loc[3, 'Valor'] = interpolacao(4)  # Preenche o valor faltante correspondente a X = 4

print(df)  # Exibe o DataFrame após o preenchimento
```

Primeiro foi criado um DataFrame com a coluna X indicando as posições dos dados e a coluna Valor contendo alguns valores ausentes, representados por np.nan. Os valores conhecidos foram separados nos arrays x e y para servirem como referência no processo de interpolação.

Depois, interp1d() foi utilizado com kind='linear' para criar uma função que calcula valores intermediários seguindo uma linha entre os pontos conhecidos. Dessa forma, entre os valores 10 e 30, o valor correspondente a X = 2 é preenchido com 20, enquanto entre 30 e 50, o valor correspondente a X = 4 é preenchido com 40.

Por fim, df.loc foi utilizado para colocar os valores calculados nas posições que estavam vazias. Assim, o DataFrame mantém todos os seus registros e os dados faltantes são estimados com base nos valores que aparecem antes e depois deles. A interpolação linear com interp1d() é apresentada como uma técnica que também pode ser utilizada para preencher dados ausentes

6. Plote dois graficos de linha em subplots diferentes: y1 = ex e y2 = log(x)

```python
import matplotlib.pyplot as plt  # Importa o módulo utilizado para criar gráficos
import numpy as np  # Importa a biblioteca NumPy

x = np.linspace(0.1, 10, 100)  # Cria 100 valores positivos entre 0.1 e 10

y1 = np.exp(x)  # Calcula y1 = e^x para cada valor de x
y2 = np.log(x)  # Calcula y2 = log(x) para cada valor de x

plt.subplot(2, 1, 1)  # Cria o primeiro subplot em uma estrutura de 2 linhas e 1 coluna
plt.plot(x, y1)  # Plota a função exponencial
plt.title("Gráfico de y1 = e^x")  # Define o título do primeiro gráfico
plt.xlabel("Eixo X")  # Define o nome do eixo horizontal
plt.ylabel("y1")  # Define o nome do eixo vertical

plt.subplot(2, 1, 2)  # Cria o segundo subplot
plt.plot(x, y2)  # Plota a função logarítmica
plt.title("Gráfico de y2 = log(x)")  # Define o título do segundo gráfico
plt.xlabel("Eixo X")  # Define o nome do eixo horizontal
plt.ylabel("y2")  # Define o nome do eixo vertical

plt.tight_layout()  # Ajusta o espaço entre os dois gráficos

plt.show()  # Exibe os subplots
```

Primeiro, np.linspace(0.1, 10, 100) foi utilizado para criar os valores de x que serão usados nas duas funções. O intervalo começa em 0.1 porque a função logarítmica np.log(x) precisa receber valores positivos. 8. Plote um gráfico de linha com intervalos de confiança (sombreamento) para a função y = sin(x).

```python
import matplotlib.pyplot as plt  # Importa o módulo utilizado para criar gráficos
import numpy as np  # Importa a biblioteca NumPy

pontos_x = np.linspace(0, 10, 100)  # Cria 100 valores entre 0 e 10

pontos_y = np.sin(pontos_x)  # Calcula o seno de cada valor de x

erro = 0.2  # Define o valor usado para criar a faixa ao redor da função

valor_maior = pontos_y + erro  # Calcula os valores acima da linha
valor_menor = pontos_y - erro  # Calcula os valores abaixo da linha

plt.plot(pontos_x, pontos_y, label='Seno')  # Cria a linha da função seno

plt.fill_between(pontos_x, valor_menor, valor_maior, alpha=0.3)  # Cria a área sombreada

plt.title("Gráfico da função seno")  # Define o título do gráfico
plt.xlabel("Valor de x")  # Define o nome do eixo horizontal
plt.ylabel("Valor de y")  # Define o nome do eixo vertical
plt.grid(True)  # Adiciona uma grade ao gráfico
plt.legend()  # Exibe a legenda

plt.show()  # Exibe o gráfico
```

Primeiro, foram criados 100 valores entre 0 e 10 para representar os pontos de x. Em seguida, np.sin() foi utilizado para calcular o valor do seno correspondente a cada um desses pontos.

Foi utilizado o valor 0.2 como margem. Esse valor é somado e subtraído dos resultados da função para criar os limites superior e inferior da área que será sombreada.

Depois, plt.plot() cria a linha da função seno e plt.fill_between() preenche o espaço entre os valores menores e maiores, formando o sombreamento ao redor da linha. O parâmetro alpha=0.3 deixa essa área parcialmente transparente, permitindo visualizar a função principal com mais facilidade.
